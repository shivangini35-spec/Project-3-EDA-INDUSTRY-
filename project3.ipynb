# Univariate Analysis

```python
df.skew(numeric_only=True).round(4)
```

annual_revenue_million shows very slight positive skewness,
indicating revenue distribution is nearly balanced with a small tendency
toward higher revenue values in some companies.
Agar strong positive skewness hota to value generally:

> 0.5 moderate skew
> 1 high skew

##

id IS NORMAL SKEWNESS
employee_count,annual_revenue_million,profit_margin_percent,founded_year   ARE POSITIVE SKEWNESS           
customer_countmarket_rating ARE NEGATIVE SKEWNESS

```python
df.kurt(numeric_only=True).round(4)
```

annual_revenue_million shows negative kurtosis,
indicating a flatter distribution with fewer extreme revenue outliers.
Most companies have revenue values spread more evenly across the range.
Kurtosis Understanding
> 0 → High peaks + more outliers (Leptokurtic)
< 0 → Flat distribution + fewer outliers (Platykurtic)
≈ 0 → Normal-like distribution

Tumhara:
−1.1871<0

👉 Hence:

flatter distribution
fewer extreme revenue values
less heavy tails

All are negative kurtosis means less outliers

```python
df['annual_revenue_million'].hist()
plt.show()
```

```python
plt.figure(figsize=(16,7))
plt.hist(df['annual_revenue_million'], bins=10, edgecolor='black', alpha=0.7)
plt.xlabel('Annual Revenue (Million)')
plt.ylabel('Frequency')
plt.title('Distribution of Annual Revenue')
plt.grid(True)
plt.show()
```

```python
sns.boxplot(x=df['annual_revenue_million'])
plt.title('Box Plot of Annual Revenue') 
plt.xlabel('Annual Revenue (Million)')
plt.grid(True)
plt.show()
```

```python
df.boxplot(column='annual_revenue_million', by='industry')
plt.title('Box Plot of Annual Revenue by Industry')
plt.suptitle('')
plt.xlabel('Industry')  
```

```python
df.boxplot(column='annual_revenue_million', by='country')
plt.title('Box Plot of Annual Revenue by country')
plt.suptitle('')
plt.xlabel('country') 
```

```python

df.boxplot(column='annual_revenue_million', by='region',figsize=(16,10))
plt.title('Box Plot of Annual Revenue by region')
plt.suptitle('')
plt.xlabel('region') 
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    x='company_name',
    y='market_rating',
    data=df
)

plt.ylim(0, 5)                    # axis 0 se 5 tak
plt.yticks([0, 1, 2, 3, 4, 5])    # ticks 0,1,2,3,4,5 par
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    x='industry',
    y='market_rating',
    data=df
)

plt.ylim(0, 5)                    # axis 0 se 5 tak
plt.yticks([0, 1, 2, 3, 4, 5])    # ticks 0,1,2,3,4,5 par
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(16,7))
plt.hist(df['market_rating'], bins=10, edgecolor='black', alpha=0.7)
plt.xlabel('Market Rating')
plt.ylabel('region')
plt.title('Distribution of Market Ratings')
plt.grid(True)
plt.show()
```

```python
df.groupby('industry')['market_rating'].max().plot(kind='bar',color ='maroon')
plt.title("Industry Performance Ratings")
plt.xlabel("Industry")
plt.ylabel("Ratings")
plt.show()
```
