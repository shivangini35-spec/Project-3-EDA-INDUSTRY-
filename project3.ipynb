#### General Rule of Thumb for Skewness
A) Between -0.5 and 0.5: Fairly symmetric (Normal-ish distribution).

B) Between -1 and -0.5 or 0.5 and 1: Moderately skewed.

C) Less than -1 or greater than 1: Highly skewed.

* Annual_revenue_million shows very slight positive skewness,
indicating revenue distribution is nearly balanced with a small tendency
toward higher revenue values in some companies.

* employee_count,annual_revenue_million,profit_margin_percent,founded_year ARE POSITIVE SKEWNESS,           
customer_countmarket_rating ARE NEGATIVE SKEWNESS

```python
df[numeric_cols].kurt(numeric_only=True).round(4)
```

```python
labels = numeric_cols  
skewness = df[numeric_cols].skew(numeric_only=True).round(2).to_list()
kurtosis = df[numeric_cols].kurt(numeric_only=True).round(2).to_list()

# 2. Set up the positions for the bars
x = np.arange(len(labels))
width = 0.35  # Width of each individual bar

# 3. Create the plot
fig, ax = plt.subplots(figsize=(10, 6))

# Plot Skewness bars (shifted to the left)
rects1 = ax.bar(x - width/2, skewness, width, label='Skewness', color='#1f77b4')

# Plot Kurtosis bars (shifted to the right)
rects2 = ax.bar(x + width/2, kurtosis, width, label='Kurtosis', color='#ff7f0e')

# 4. Add formatting and labels
ax.set_ylabel('Value Scale')
ax.set_title('Skewness and Kurtosis for Numerical Values')
ax.set_xticks(x)
ax.set_xticklabels(labels)
ax.axhline(0, color='black', linewidth=0.8, linestyle='--') # Baseline at zero
ax.grid(axis='y', linestyle='--', alpha=0.3)
ax.legend(loc='upper right',bbox_to_anchor=(1.20, 1))

# 5. Add value labels on top of the bars for clarity
ax.bar_label(rects1, padding=3, fmt='%.2f')
ax.bar_label(rects2, padding=3, fmt='%.2f')

plt.tight_layout()
plt.show()
```

Annual_revenue_million shows negative kurtosis, indicating a flatter distribution with fewer extreme revenue outliers.
Most companies have revenue values spread more evenly across the range.
Kurtosis Understanding
> 0 → High peaks + more outliers (Leptokurtic)
< 0 → Flat distribution + fewer outliers (Platykurtic)
≈ 0 → Normal-like distribution


−1.1871<0

👉 Hence:

flatter distribution
fewer extreme revenue values
less heavy tails

All are negative kurtosis means less outliers.

```python
Industry= df['industry'].value_counts()

def func(pct):
    count = int(round(pct/100 * Industry.sum()))
    return f'{count}\n({pct:.1f}%)'

plt.figure(figsize=(8,8))

plt.pie(
    Industry,
    labels=Industry.index,
    autopct=func
)

plt.title("Industry Distribution")

plt.show()
```


### Separate Insights

* **Absolute Frequency Analysis:** This shows the **exact count** of data points (rows) present for each unique category in the `industry` column.
* **Uniform Dataset Size:** It proves that subset size for every single industry is perfectly identical. Adding them together ($3000 \times 5$) immediately reveals total dataset size is exactly **15,000 rows**. 

* **Relative Proportion Analysis:** By using `normalize=True`, I converted absolute numbers into fractions of a whole, and multiplying by `100` converts those fractions into **percentages (%)**.
* **Perfect Class Symmetry:** This explicitly highlights that each industry commands an exact **20% share** of the total dataset.

# IQR Method for outliers detection

```python
Q1 = df['annual_revenue_million'].quantile(0.25)
Q3 = df['annual_revenue_million'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['annual_revenue_million'] < lower) |
              (df['annual_revenue_million'] > upper)]

print(outliers)
```

```python
Q1 = df['employee_count'].quantile(0.25)
Q3 = df['employee_count'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df['employee_count'] < lower) |
    (df['employee_count'] > upper)
]

print(outliers)
```

```python
Q1 = df['customer_count'].quantile(0.25)
Q3 = df['customer_count'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df['customer_count'] < lower) |
    (df['customer_count'] > upper)
]

print(outliers)
```

No significant outliers were detected using the IQR method,
indicating a relatively balanced distribution.
